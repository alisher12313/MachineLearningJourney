```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('teams.csv')
df.head()
```

```python
df.isna().sum()
```

```python
train, test = train_test_split(df, test_size=.2, random_state=42)
```

```python
predictors = ['athletes', 'events']
target = ['medals']
```

```python
X = train[predictors].copy()
y = train[target].copy()
```

```python
X.describe()
```

```python
X = (X - X.mean()) / X.std()
```

```python
X.describe()
```

```python
X['intercept'] = 1
X = X[['intercept'] + predictors]
```

```python
alpha = 2
I = np.eye(X.shape[1])
```

```python
I
```

```python
I[0 ,0] = 0
```

```python
penalty = alpha * I
```

```python
B = np.linalg.inv(X.T @ X + penalty) @ X.T @ y
```
