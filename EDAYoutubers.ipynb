```python
pd.set_option('display.float_format', '{:.0f}'.format)
```

```python
category_subscribers_mean = youtube_df.groupby('category')['subscribers'].mean().reset_index()
category_subscribers_mean.sort_values('subscribers', ascending=False, inplace=True)
```

```python
sns.barplot(data=category_subscribers_mean, x='subscribers', y='category')
```

```python
channel_types_count = youtube_df['channel_type'].value_counts().reset_index()
channel_types_count
```

```python
sns.barplot(data=channel_types_count, y='channel_type', x='count')
```
