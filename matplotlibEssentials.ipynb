```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names()
```

```python
crash_df = sns.load_dataset('car_crashes')
crash_df.head()
```

Distribution Plot

```python
sns.displot(crash_df['not_distracted'], kde=False, bins=25)
```

Joint plot


```python
sns.jointplot(x='speeding', y='alcohol', data=crash_df, kind='reg')
```

kde plot

```python
sns.kdeplot(crash_df['alcohol'])
```

pair plot

```python
sns.pairplot(crash_df)
```

```python
tips_df = sns.load_dataset("tips")
tips_df.head()
```

```python
sns.pairplot(tips_df, hue='sex', palette='Blues')
```

```python
sns.rugplot(tips_df['tip'])
```

Styling

```python
sns.set_style("ticks")

plt.figure(figsize=(8, 4))

sns.set_context('paper', font_scale=1.4)

sns.jointplot(x=crash_df['speeding'], y=crash_df['alcohol'], kind='reg')
```

Bar plot

```python
sns.barplot(x=tips_df['sex'], y=tips_df['total_bill'], estimator=np.mean)
```

```python
sns.countplot(x=tips_df['sex'])
```

```python
sns.boxplot(x='day', y='total_bill', data=tips_df, hue='sex')
plt.legend(loc=0)
```

```python
sns.violinplot(x='day', y='total_bill', data=tips_df, hue='sex')
```

```python
sns.stripplot(x='day', y='total_bill', data=tips_df, hue='sex', jitter=True, dodge=True)
```

```python
sns.swarmplot(x='day', y='total_bill', data=tips_df)
```
