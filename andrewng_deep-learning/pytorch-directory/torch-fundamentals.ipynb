```python
torch.matmul(tensor_operation2, tensor_operation2.T), tensor_operation2 @ tensor_operation2.T
#to get correct multiplication we must follow 2 rules:
# inner dimensions must match
# the output of mult will result in outer dimensions size:
# 2, 3 and 2, 3 -> won't work
# 2, 3 and 3, 2 -> will work, new output: 2, 2
# 3, 2 and 2, 3 -> will work, new output: 3, 3
```

## Reshaping, squeezing, stacking, unsqueezing:
* Reshaping - reshapes an input to a defined shape
* View - return a view of an input tensor of certain shape but keep the same memory as the original tensor
* Stacking - combine multiple tensors on top of each other `(vstack)` or side by side `(hstack)`
* Squeeze - remove all 1 dimensions from tensor
* Unsqueeze - adds 1 dimension to tensor
* Permute - return a view of the input with dimensions permuted in a certain way

```python
x2 = torch.arange(1, 13, dtype=torch.float32)
x2, x2.shape
```

# Reproducibility (trying to take random out of random)

In short, how a neural network learns:

start with random numbers -> tensor operations -> update random numbers to try and make them better representations of the data -> again -> again -> again...

That loop is training. Every weight starts as a random number (that torch.randn() * 0.01 from earlier), the forward pass runs tensor operations on it, the loss tells you how wrong it was, and backprop nudges the random numbers to be slightly less random — slightly more tuned to the actual data. Repeat thousands of times, and "random" gradually becomes "learned."

The problem: because it starts random and gets nudged by random data shuffling too, every run is different by default — different starting weights, different training order. Fine for real training, bad for debugging or comparing two versions of your code, because you can't tell if a change in results came from your code or from a new roll of random numbers.

torch.manual_seed(42) is how you take the "random" back out of random — pin the pseudo-random generator to always produce the same sequence, so reruns are identical and comparisons are fair.
