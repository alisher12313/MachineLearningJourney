```python
import time

a = np.random.rand(1_000_000)
b = np.random.rand(1_000_000)

tic = time.time()
c = np.dot(a, b)
toc = time.time()

print(1000*(toc - tic))

c = 0
tic = time.time()
for i in range(1_000_000):
    c += a[i] + b[i]

toc = time.time()

print(1000*(toc - tic))
```
