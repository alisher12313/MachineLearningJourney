__Filtering__


```python
ages = np.array([
    [21, 17, 19, 20, 16, 30, 18, 65],
    [39, 22, 15, 99, 18, 19, 20, 21]
])
```

```python
teenagers = ages[ages < 18]
teenagers
```

```python
adults = ages[(ages >= 18) & (ages < 65)]
adults
```

```python
np.sort(adults)
```

```python
adults
```

```python
seniors = ages[ages >= 65]
seniors
```

```python
evens = ages[ages % 2 == 0]
evens
```

```python
odds = ages[ages % 2 != 0]
odds
```

```python
adults = np.where(ages >= 18, ages, -1)
adults
```
